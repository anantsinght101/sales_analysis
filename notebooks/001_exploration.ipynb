```python
#To see if the duplicated rows and the registers with the same codes and quantities are the same
duplicados = df[df.duplicated(keep=False)]
duplicados.groupby(['Invoice', 'StockCode', 'Quantity']).size()
```

```python
# Repeated combinations with Invoice, StockCode y Quantity
conteo = df.groupby(['Invoice', 'StockCode', 'Quantity']).size()

# Repited combinations that aren't exact duplicates
grupos_no_exactos = conteo[conteo > 1].drop(
    index=duplicados.groupby(['Invoice', 'StockCode', 'Quantity']).size().index,
    errors='ignore'
)

grupos_no_exactos
```

```python
df.set_index(['Invoice', 'StockCode', 'Quantity']).loc[grupos_no_exactos.index]
```

6825 rows were identified as exact duplicates, equivalent to 1.31% of the dataset. Some lines of the same Invoice (check) have exactly the same value in every column. It'll be conserved only one register and the copies will be droped.
